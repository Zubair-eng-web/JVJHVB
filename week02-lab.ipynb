```python
import cv2
from pathlib import Path

# Read the input image from the images folder.
image_dir = Path('images')
input_path = image_dir / 'input.jpg'
image = cv2.imread(str(input_path))

if image is None:
    raise FileNotFoundError(f'Could not load image from {input_path}.')

# Convert the color image to grayscale.
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Make a black-and-white mask using 128 as the cutoff.
_, binary_mask = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY)

# Blur the grayscale image with a 5x5 filter.
blurred = cv2.blur(gray, (5, 5))

# Save the processed images in the images folder.
cv2.imwrite(str(image_dir / 'output_mask.jpg'), binary_mask)
cv2.imwrite(str(image_dir / 'output_blur.jpg'), blurred)
cv2.imwrite(str(image_dir / 'output_gray.jpg'), gray)
```

```python
import numpy as np


def adjust_brightness_contrast(image_array, alpha=1.0, beta=0):
    # Use float values to avoid overflow during the calculation.
    adjusted = image_array.astype(np.float32) * alpha + beta
    # Keep pixel values in the 0-255 range and return uint8 data.
    return np.clip(adjusted, 0, 255).astype(np.uint8)


# alpha changes contrast; beta adds or subtracts brightness.
gray_example = np.array([[100, 150], [200, 50]], dtype=np.uint8)
result = adjust_brightness_contrast(gray_example, alpha=1.5, beta=-20)
print(result)
```
